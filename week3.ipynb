Electricity bill


```python
name = input("Enter customer name: ")
meter = input("Enter meter number: ")
old = int(input("Enter old reading: "))
new = int(input("Enter new reading: "))

units = new - old

if units <= 50:
    bill = units * 2.30
elif units <= 200:
    bill = units * 4
elif units <= 400:
    bill = units * 5
else:
    bill = units * 5.40

print("\nElectricity Bill")
print("Customer Name:", name)
print("Meter Number:", meter)
print("Units Consumed:", units)
print("Bill Amount: Rs.", bill)
```

Factorial of number

```python
n = int(input("Enter a number: "))
fact = 1

for i in range(1, n + 1):
    fact = fact * i

print("Factorial =", fact)
```
