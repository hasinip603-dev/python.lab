```python
#Program on creating class and object
class Student:
  def __init__(self,name,marks):
    self.name=name
    self.marks=marks
  def display(self):
    print(self.name,self.marks)
s1=Student("Sri",90)
s2=Student("Ravi",96)
s1.display()
s2.display()
#__init__=built in constructor called initiator
#self:at that instance
```

Write a python program to create a class that represents a shape include methods to calculate are and perimeter.Implement subclasses for different shapes like triangle,square,and circle.



```python
#Program illustarting Inheritance and overwritting methods
class Shape:
  def area():
    pass
  def perimeter():
    pass
class Circle(Shape):
  def __init__(self,r):
    self.r=r
  def area(self):
    return 3.14*self.r*self.r
  def perimeter(self):
    return 2*3.14*self.r
class Square(Shape):
  def __init__(self,side):
    self.side=side
  def area(self):
      return self.side*self.side
  def perimeter(self):
      return self.side*4
class Triangle(Shape):
  def __init__(self,a,b,c,h):
    self.a=a
    self.b=b
    self.c=c
    self.h=h
  def area(self):
      return 0.5*self.b*self.h
  def perimeter(self):
      return self.a+self.b+self.c
C1=Circle(10)
print("The area of circle :", C1.area())
print("The perimeter of circle:",C1.perimeter() )
S1=Square(10)
print("The area of square:",S1.area())
print("The perimeter of square:",S1.perimeter())
T1=Triangle(10,10,10,10)
print("The area of triangle:",T1.area())
print("The perimeter of triangle:",T1.perimeter())
```
