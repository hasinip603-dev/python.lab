```python
program that checks whether the palindrome or not using functions
```

```python
def palindrome(n):
  original = n
  reverse =0
  while(n>0):
    k=n%10
    reverse=(reverse*10)+k
    n=n//10
  if(reverse==original):
    print("palindrome")
  else:
    print("not palindrome")
n=int(input())
palindrome(n)
```

```python
#example for variable length arguments
def add(*num):
 print(sum(num))
add(1,2,3,4,5)
```

```python
#FUNCTION THAT RETURNS MORE THAN ONE VALUE
def multiple_values(*num):
  print(num)
  s=sum(num)
  c=len(num)
  avg=s/c
  return s,c,avg
a,b,d=multiple_values(1,2,3,4,5)
print(f"Sum is {a},Count is {b} and Average is {d}")
```
