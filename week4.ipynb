operations of list in python

```python
L=[10,20,30,40,50]
print(L)
print("indexing")
print(L[0])
print(L[-1])
print("slicing")
print(L[1:4])
print(L[:3])
print(L[2:])
print(L[::-1]) #PRINT ELEMENTS IN REVERSE ORDER
print("concatenation")
print(L+[60,70])
print("repetation")
print(L*2)
print("membership")
print(25 in L)
print(10 in L)
print("iteration")
for i in L:
  print(i)
```

Program: Print the sum of list elements without using sum() function

```python
L=list(map(int,input().split()))
sum=0
for i in L:
  sum=sum+i
print(sum)
```

Creating a list in run time 1.Using map function 2.List comprehension 3.Normal approach

```python
#NORMAL APPROACH WITHOUT FUNCTIONS
a=[]
n=int(input())
for i in range(n):
  a.append(int(input()))
print(a)
```

```python
#USING MAP FUNCTION
a=list(map(int,input().split()))
a=map(int,input().split())
a=input().split()
a=list(map(int,input().split("/")))
print(a)
```

```python
#LIST COMPREHENSION:It is a way of creating a new list using existing sequence
a=[int(i) for i in input().split()]
print(a)
L1=[value for value in range(10)]
print(L1)
L2=[i**2 for i in L1]
print(L2)
L3=[i for  i in L2 if (i%2==0)]
print(L3)
```
