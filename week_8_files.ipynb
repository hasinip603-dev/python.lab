```python
file=open("hasini.txt","w")
content="my name is hasini"
file.write(content)
print("file created sucessfully")
file.close()
```

```python
#when we write the content into the same filr using write it overwrites the current content
file=open("hasini.txt","w")
content2="i am currently pursuing my b.tech 2nd year"
file.write(content2)
print("file created sucessfully")
file.close()
```

```python
#for appending use append mode
file=open("hasini.txt","a")
file.write("at vishnu enginnering college")
print("file created sucessfully")
file.close()
```

```python
#FOR APPENDING in different lines use writelines
file=open("hasini.txt","w")
file.writelines("my name is hasini\n")
file.writelines("i am currently pursuing my b.tech 2nd year\n")
file.writelines("at vishnu enginnering college")
print("file created sucessfully")
file.close()
```

```python
#for reading use read mode
with open ("hasini.txt","r")as file:
  data=file.read()
  print(data)
```

```python
#arguments as n reads n characters
with open ("hasini.txt","r") as file:
 data=file.readline(10)
 print(data)
```

```python
#to read all the sentences as a list elements by using readlines method
with open("hasini.txt","r")as file:
  list1=file.readlines()
  print(list1)
```

```python
#to know the position of cursor we have tell method
file=open("hasini.txt","r")
file.readline()
print(file.tell())
```

```python
#to chage the position of cursor we have seek method
file=open("hasini.txt","r")
file.seek(10)
```
