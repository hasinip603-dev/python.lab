```python
#operations in tuples
t1=(7,9,10,11,12)
t2=(5,2,8,16)
print(t1+t2) #concatination
print(t2*2) #repetation
for i in t1:
  print(i)
print(t1[0:3]) #slicing
print(t2[::-1]) #reversing
print(t1[2]) #indexing
```

```python
#functions in tuples
s=(20,2,30,3,40,12,31,4,28)
print(len(s))
print(max(s))
print(min(s))
print(sorted(s))
```

```python
#methods in tuples
p=(2,23,12,20,30,41,1,50,1)
print(p.count(1))
print(p.index(20))
```

```python
#zipping
a1=(2,3,5)
a2=(6,8,9)
print(tuple(zip(a1,a2)))
```

```python
#creating tuple
k=tuple(map(int,input("enter the tuple elements").split()))
print(k)
```

```python
#using comprehension
c=tuple(int(i) for i in input().split())
print(c)
```

```python
#write a program
k=('hasini',19,'vinukonda','vishnu')
v=('vyshu',18,'narsaropeta','nec')
print(k+v)
```

```python
#write a program to count the number of vowels in a sting (NO control #flow allowed)
s=str(input("enter a string:"))
s1=s.lower()
vowels_count=s1.count("a")+s1.count("e")+s1.count("i")+s1.count("o")+s1.count("u")
print(vowels_count)
```

```python
#operations in sets
A={1,2,3,4,5}
B={4,5,6,7,8}
print(A|B)
print(A&B)
print(A-B)
print(A^B)
```

```python
#update
A={1,2,3,4,5}
B={4,5,6,7,8}
A.update(B)
print(A)
```
