```python
#operations on sets
a={1,2,3}
b={"hasi",4,7,6,3,2}
print(a&b) #intersection
print(a|b) #union
print(a-b) #difference
print(b-a) #difference
print(a^b) #Symmetric difference
print(a)
print(b)
```

```python
#Conversion of list into set and viceversa
L1=list(map(int,input().split()))
L2=list(map(int,input().split()))
print(L1)
print(L2)
S1=set(L1)
S2=set(L2)
print(S1)
print(S2)
print(len(S1),len(S2))
```

```python
#If a number having same digits print YES otherwise NO
number=int(input())
s=str(number)
m=set(s)
if(len(m)>1):
  print("NO")
else:
  print("YES")
```

```python
#To change the original set
c={11,22,33}
a={1,2,3,33}
print(a.union(c))
print(a.difference(c))
print(a.intersection(c))
print(a.intersection_update(c))
print(a)
```

```python
#find no of duplicates in list
L=list(map(int,input().split()))
K=set(L)
print(len(L)-len(K))
```

```python
#number of elements repeated
L1=[i for i in input().split()]
L2=[]
for i in L1:
  if L1.count(i)>1:
    L2.append(i)
L3=set(L2)
print(len(L3))
```

```python
#Write a python program that prints common elements between two lists
L1=[i for  i in input().split()]
L2=[i for i in input().split()]
p=set(L1)
q=set(L2)
print(p & q)
```

```python
#Write a python program that find missing number between 1 and n using a set
n=int(input())
L1=list(map(int,input().split()))
L2=[]
for i in range(1,n+1):
  L2.append(i)
p=set(L1)
q=set(L2)
print(q-p)
```

```python
#Write a python program that prints only unique elements in a list
L1=[i for i in input().split()]
L2=[]
for i in L1:
  if L1.count(i)==1:
    L2.append(i)
print(L2)
```
